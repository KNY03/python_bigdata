````python
# 문제 1 
# 1부터 50까지의 숫자 중 짝수만 출력하는 프로그램을 작성하시오.

for i in range(1, 51):
    if i % 2 == 0:
        print(i)
````

````python
# 문제 2 
# 사용자로부터 숫자를 입력받아 해당 숫자의 구구단을 출력하는 프로그램을 작성하시오.

num = int(input("숫자를 입력하세요: "))
for i in range(1, 10):
    print(f"{num} x {i} = {num * i}")
````

````python
# 문제 3 
# 다음과 같은 패턴을 출력하는 프로그램을 작성하시오.

# **출력:**
# ```
# *
# **
# ***
# ****
# *****

num = int(input("숫자를 입력하세요: "))

for i in range(1, num + 1):
    print('*' * i)
````

````python
# 문제 4
# 두 개의 숫자를 입력받아 두 수 사이의 모든 정수의 합을 반환하는 함수 sum_between(a, b)를 작성하시오.
# (a와 b 포함)

def sum_between(a, b):
    return sum(range(a, b + 1))

a = int(input("첫 번째 숫자를 입력하세요: "))
b = int(input("두 번째 숫자를 입력하세요: "))
print(sum_between(a, b))
````

````python
# 문제 5
# 가변 인자를 받아 다음 정보를 계산하는 함수 calculate_stats(*numbers)를 작성하시오.

# 평균 (소수점 2자리)
# 최댓값
# 최솟값
# 개수

def calculate_stats(*numbers):
    count = len(numbers)
    maximum = max(numbers)
    minimum = min(numbers)
    average = round(sum(numbers) / count, 2)
    return average, maximum, minimum, count

nums = input("숫자들을 공백으로 구분하여 입력하세요: ").split()
nums = [int(num) for num in nums]
avg, max_val, min_val, cnt = calculate_stats(*nums)
print(f"평균: {avg}, 최댓값: {max_val}, 최솟값: {min_val}, 개수: {cnt}")
````

````python
todos = []  # 할 일 목록


def add_todo():
    """1. 할 일 추가"""
    # 할 일 입력 받기
    text = input("추가할 할 일: ").strip()

    # 빈 문자열이면 안내문 출력
    if text == "":
        print("빈 내용은 추가할 수 없습니다.\n")
        return
    
    # todos에 추가
    todos.append({"text": text, "done": False})

    # 완료 메시지 출력
    print("추가되었습니다.\n")


def show_todos():
    """2. 할 일 목록 조회"""
    # todos가 비어있으면 비어있다고 출력
    if len(todos) == 0:
        print("\n할 일이 없습니다.\n")
        return
    
    # 번호, 완료 여부, 할 일 출력
    print("\n------ Todo List ------")
    for i in range(len(todos)):
        status = "완료" if todos[i]["done"] else "미완료"
        print(f"{i+1}. [{status}] {todos[i]['text']}")
    print("--------------------\n")


def complete_todo():
    """3. 할 일 완료 처리"""
    # todos가 비어있으면 안내 출력
    if len(todos) == 0:
        print("완료 처리할 할 일이 없습니다.\n")
        return

    # 완료할 번호 입력 받기
    num = input("완료할 번호: ").strip()
    
    idx = int(num) - 1
    if idx < 0 or idx >= len(todos):
        print("번호가 올바르지 않습니다.\n")
        return

    # todos[idx]["done"] = True 처리
    # 완료 메시지 출력
    todos[idx]["done"] = True
    print("완료 처리되었습니다.\n")



def delete_todo():
    """4. 할 일 삭제"""
    # todos가 비어있으면 안내 출력
    if len(todos) == 0:
        print("삭제할 할 일이 없습니다.\n")
        return
    
    # 삭제할 번호 입력 받기
    num = input("삭제할 번호: ").strip()

    # remove
    idx = int(num) - 1
    if idx < 0 or idx >= len(todos):
        print("번호가 올바르지 않습니다.\n")
        return


    removed = todos.pop(idx)


    # 삭제 메시지 출력
    print("삭제되었습니다:", removed["text"], "\n")



def main():
    # 프로그램 반복 실행(메뉴)
    while True:
        print("1) 추가  2) 목록  3) 완료  4) 삭제  0) 종료")
        menu = input("메뉴 선택: ").strip()


        if menu == "1":
            # add_todo() 호출
            add_todo()

        elif menu == "2":
            # show_todos() 호출
            show_todos()

        elif menu == "3":
            # complete_todo() 호출
            complete_todo()

        elif menu == "4":
            # delete_todo() 호출
            delete_todo()

        elif menu == "0":
            # 종료 메시지 출력 후 break
            print("종료합니다.\n")
            break

        else:
            print("메뉴를 다시 선택하세요.")


if __name__ == "__main__":
    main()
````
